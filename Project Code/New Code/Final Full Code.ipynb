```python
import os

# Function to explore dataset structure
def explore_dataset(folder_path):
    for tumor_type in os.listdir(folder_path):  # Loop through each tumor category
        tumor_path = os.path.join(folder_path, tumor_type)
        if os.path.isdir(tumor_path):
            num_images = len(os.listdir(tumor_path))
            print(f"Tumor Type: {tumor_type}, Number of Images: {num_images}")

# Explore Training dataset
print("Training Dataset Overview:")
explore_dataset(os.path.join(dataset_path, "Training"))

# Explore Testing dataset
print("\nTesting Dataset Overview:")
explore_dataset(os.path.join(dataset_path, "Testing"))
```

```python
import cv2
import numpy as np

# Define image size
IMAGE_SIZE = (224, 224)

# Function to load and preprocess images
def load_images(folder_path):
    images, labels = [], []
    for tumor_type in os.listdir(folder_path):
        tumor_path = os.path.join(folder_path, tumor_type)
        if not os.path.isdir(tumor_path):
            continue
        for img_file in os.listdir(tumor_path):
            img_path = os.path.join(tumor_path, img_file)
            img = cv2.imread(img_path)
            img = cv2.resize(img, IMAGE_SIZE)
            img = img / 255.0  # Normalize pixel values to [0,1]
            images.append(img)
            labels.append(tumor_type)
    return np.array(images), np.array(labels)

# Load training images
train_images, train_labels = load_images(os.path.join(dataset_path, "Training"))
print("Training images shape:", train_images.shape)
print("Training labels shape:", train_labels.shape)
```
